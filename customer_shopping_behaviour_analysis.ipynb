# Data Cleaning

### 1. Setup
#### Importing the libraries needed for data loading, cleaning, and transformation.

```python
##Package Import

import pandas as pd
import numpy as np
import datetime as dt
```

### 2. Load the Dataset
#### Reading the raw customer shopping behavior data and previewing the first few rows to get an initial sense of the structure.

```python
#Loading the dataset using pandas

cust = pd.read_csv('customer_shopping_behavior_DATA.csv')
cust.head()
```

### 2.1 Initial Exploration
#### Checking the number of unique values in each column to get a quick sense of which fields are categorical, which are near-continuous, and where potential data quality issues might be hiding.

```python
cust.nunique()
```

#### Viewing summary statistics for the numeric columns to spot any obvious outliers or inconsistencies.

```python
cust.describe()
```

#### Checking data types and non-null counts for each column to identify missing values.

```python
cust.info()
```

#### Checking for duplicate rows in the dataset.

```python
cust.duplicated().sum()
```

### 3. Column Name Standardization
#### Cleaning up column headers — stripping whitespace, converting to lowercase, replacing spaces with underscores, and removing parentheses — so column names are consistent and easier to reference going forward.

```python
#Renaming columns according to snake casing for better readability

cust = cust.rename(columns = lambda x: x.strip().lower().replace(' ','_').replace('(','').replace(')',''))
cust.head()
```

### 4. Handling Missing Values
#### Checking and filling missing values

#### Filling missing `review_rating` values with the column mean

```python
# Checking if missing data or null values are present in the dataset

cust.isnull().sum()
```

```python
cust.review_rating = cust.review_rating.fillna(cust.review_rating.mean().round(2))
```

```python
cust.isnull().sum()
```

### 5. Removing Redundant Columns
#### Checking whether `discount_applied` and `promo_code_used` carry the same information by counting mismatches between them.

```python
sum(cust['discount_applied'] != cust['promo_code_used'])
```

#### Since the two columns matched with zero mismatches, `discount_applied` is redundant and is dropped to avoid duplicate information.

```python
#Dropping discount_applied column

cust.drop(columns='discount_applied', inplace=True)
```

### 6. Feature Engineering: Purchase Frequency
#### Viewing the unique categories in `frequency_of_purchases` before converting them into a numeric, analysis-friendly format.

```python
cust.frequency_of_purchases.unique()
```

#### Mapping each purchase frequency category to an approximate number of days (e.g. Weekly → 7, Monthly → 30, Annually → 365). This turns a categorical field into a numeric one that can be used directly in calculations and SQL queries later.

```python
#Creating new column for purchase frequency in days

frequency_map = {'Fortnightly':14, 'Weekly':7, 'Annually':365, 'Quarterly':90, 'Bi-Weekly':14, 'Monthly':30, 'Every 3 Months':90}

cust['purchase_frequency_days'] = cust['frequency_of_purchases'].map(frequency_map)
cust.head()
```

#### Renaming `previous_purchases` to `repeat_purchase` for clarity, since the column represents the count of prior purchases made by a customer.

```python
#Renaming previous_purchases to repeat_purchase_count for better understanding

cust.rename(columns={'previous_purchases':'repeat_purchase_count'},inplace=True)
```

### 7. Feature Engineering: Age Groups
#### Checking the range of unique ages present in the data before binning them into groups.

```python
cust.age.unique()
```

#### Creating an `age_group` column by binning ages into 5-year buckets (16–20, 21–25, etc.) and inserting it next to the `age` column. Age and age_group are kept as separate columns so both granular and grouped analysis remain possible downstream.

```python
#Create a new column age_group

bins = [15,20,25,30,35,40,45,50,55,60,65,70]
labels = ['16-20','21-25','26-30','31-35','36-40','41-45','46-50','51-55','56-60','61-65','66-70']
age_group= pd.cut(cust.age,bins= bins, labels=labels)
cust.insert(1,'age_group', age_group)
```

#### Final preview of the cleaned dataset before exporting it for the SQL stage.

```python
cust.head()
```

```python
#Exporting the clean dataset to .csv
cust.to_csv(r"D:\02_Cleaned Data.csv",index=False)
```

- **Imputed** missing `review_rating` values using the column mean.
